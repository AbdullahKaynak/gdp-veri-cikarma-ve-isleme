```python
import pandas as pd
import numpy as np

url = "https://en.wikipedia.org/wiki/List_of_countries_by_GDP_%28nominal%29"
```

## 1-) pandas ile web sayfası tablolarını çıkar. ve ilgili 3. tabloyu al 

```python
df = pd.read_html(url, storage_options={"User-Agent": "Mozilla/5.0"})[2]
df
```

## Sütun başlıklarını sütun numaralarıyla değiştirin

```python
df.columns = range(len(df.columns))
df.columns = [0, 1, 2, 3]
df
```

## 0 ve 2. indeksli sütunları saklayın (ülke adı ve IMF tarafından belirtilen GSYİH değeri)

```python
df = df.drop(columns=[1,3])
df
```

## Dünyanın en büyük 10 ekonomisini temsil eden, 1 ila 10. indeks aralığındaki satırları saklayın. 

```python
df[1:11]
```

## Sütun adlarını "Country" ve "GDP (Million USD)" olarak atayın

```python
# df2 = df.copy()
# df2.columns = ["Country", "GDP (Million USD)"]
# df2

df.columns = ["Country", "GDP(Million USD)"]
df.head()
```

### DataFrame'in GDP sütununu, Milyon USD cinsinden olan değeri Milyar USD'ye dönüştürerek değiştirin. Değeri virgülden sonra iki basamağa yuvarlamak için NumPy kütüphanesinin round() metodunu kullanın. DataFrame'in başlığını "GDP (Billion USD)" olarak değiştirin.

```python
## 'GDP (Million USD)' sütununun veri türünü tam sayıya (integer) dönüştürün.
#  astype() yöntemini kullanın.

df2 = df.copy()
df2['GDP(Million USD)'] = pd.to_numeric(df2['GDP(Million USD)'], errors='coerce').fillna(0).astype(int)

## Milyon USD cinsinden GSYİH değerini Milyar USD'ye dönüştürün.
df2["GDP(Million USD)"] = pd.to_numeric(df2["GDP(Million USD)"], errors="coerce").fillna(0)/1000
df2

# Değeri virgülden sonra iki basamağa yuvarlamak için numpy.round() yöntemini kullanın.

df3 = df2.copy()
df3["GDP(Million USD)"] = np.round(df3[["GDP(Million USD)"]], 2)
df3

# Sütun başlığını 'GDP (Million USD)' yerine 'GDP (Billion USD)' olarak değiştirin.

df3 = df3.rename(columns={"GDP(Million USD)": "GDP(BİLLİON USD)"})
df3
```

```python
# DataFrame'i "Largest_economies.csv" adlı CSV dosyasına yükleyin

df3.to_csv("Largest_economies.csv")
```
